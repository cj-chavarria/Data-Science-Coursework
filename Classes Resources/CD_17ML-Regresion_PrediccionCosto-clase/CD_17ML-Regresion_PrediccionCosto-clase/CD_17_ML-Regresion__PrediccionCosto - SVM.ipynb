#### Pipeline

La función pipeline permite juntar varios trasformaciones secuenciales que se requieren ejecutar a un conjunto de datos.

**Pipeline(steps=[('titulo trasnformación1', trasnformación1), ('titulo trasnformación2', trasnformación2),...])**

#### ColumnTransformer


ColumnTransformer es una clase de scikit-learn que  permite aplicar transformaciones de preparación de datos de forma selectiva.

Por ejemplo,  permite aplicar una transformación específica o secuencia de transformaciones solo a las columnas numéricas, y una secuencia separada de transformaciones solo a las columnas categóricas.

Para usar ColumnTransformer, debe especificar una lista de transformadores.

Cada transformador es una tupla de tres elementos que define el nombre del transformador, la transformación a aplicar y los índices de columna a los que aplicar. Por ejemplo: **(Nombre, objeto, columnas)**

```python
# Definir los transformadores
transformacion = ColumnTransformer([('num', transformacion_num, lista_num)])
```

# Hyperparámetros clave para `SVR` (Support Vector Regressor, scikit-learn)

El modelo **SVR** (Support Vector Regression) busca ajustar una función que minimiza los errores fuera de un margen de tolerancia (ε), utilizando kernels para modelar relaciones lineales o no lineales.

---

## 1. `kernel`
- **Qué es:** tipo de función que transforma los datos al espacio de mayor dimensión.
- **Efecto:** define la forma del límite de regresión (lineal o no lineal).
- **Probar:**
  - `"linear"` → relaciones simples y proporcionales.  
  - `"poly"` → relaciones polinómicas (ajusta con `degree`).  
  - `"rbf"` → relaciones no lineales suaves (más común).  
  - `"sigmoid"` → menos usado, comportamiento tipo red neuronal.

---

## 2. `C`
- **Qué es:** parámetro de penalización del error.
- **Efecto:** controla el equilibrio entre el ajuste del modelo y la simplicidad.
  - Alto `C` → menor tolerancia a errores → posible sobreajuste.  
  - Bajo `C` → más regularización → posible subajuste.
- **Probar:** `0.1, 1, 10, 100, 1000`

---

## 3. `epsilon`
- **Qué es:** margen de tolerancia dentro del cual los errores no se penalizan.
- **Efecto:** controla la sensibilidad del modelo a pequeños errores.
  - Mayor `epsilon` → menos sensible → más suave.  
  - Menor `epsilon` → más preciso → puede sobreajustar.
- **Probar:** `0.01, 0.05, 0.1, 0.2, 0.5`

---

## 4. `gamma` *(solo aplica a kernels `"rbf"`, `"poly"`, `"sigmoid"`)*
- **Qué es:** define la influencia de cada punto de entrenamiento.
- **Efecto:** controla la curvatura de la superficie de decisión.
  - Alto `gamma` → curvas más ajustadas → posible sobreajuste.  
  - Bajo `gamma` → curvas más suaves → posible subajuste.
- **Probar:** `"scale"`, `"auto"`, `0.01, 0.1, 1, 10`

---

## 5. `degree`
- **Qué es:** grado del polinomio en el kernel `"poly"`.
- **Efecto:** solo tiene efecto si `kernel='poly'`.
- **Probar:** `2, 3, 4, 5`


```python
#Aplicar el modelo RandomForestRegressor
from sklearn.svm import SVR
```

```python
#1. Nombrar una variable para nuestro modelo
SVM_regression = SVR()
```

```python
SVM_regression.get_params()
```

```python
#2. Definir el Pipeline : Transformación  + modelo

modelo_SVM = Pipeline([('transformacion', transformacion),
                       ('SVM_regression', SVM_regression)])
```

```python
# Utilizamos la función GridSearchCV para encontrar los valores optimos
from sklearn.model_selection import GridSearchCV
```

```python
#Definición de posibles parametros
C = [100, 1000, 10000]
epsilon = [0.01, 0.2]
kernel = ['rbf', 'linear']
gamma = ['scale', 0.1, 1]

parametros = {
    'SVM_regression__C': C,
    'SVM_regression__epsilon': epsilon,
    'SVM_regression__kernel': kernel,
    'SVM_regression__gamma': gamma
}
```

```python
#Definicion del grid de busqueda
grid_SVM = GridSearchCV(modelo_SVM, param_grid=parametros, cv=5,
                        scoring='neg_mean_squared_error', n_jobs=-1, verbose=2)
```

```python
# Ajuste del modelo usando todas las posibles combinaciones de parametros 
grid_SVM.fit(X,y)
```

```python
# Mostrar la mejor combinación
print( grid_SVM.best_params_ )
```

**Encontramos los nuevos valores de la media y desviación estándar**


```python
#1 Ajustar el modelo con los nuevos parametros 

SVM_regression_tunned = SVR(C=10000, epsilon=0.2, gamma='scale', kernel='linear')
```

```python
#2. Definir el Pipeline : Transformación  + modelo
modelo_SVM_tunned = Pipeline([('transformacion', transformacion),
                              ('SVM_regression_tunned', SVM_regression_tunned)])
```

```python
#3. Ajustar el modelo
modelo_SVM_tunned.fit(X,y)
```

```python
from sklearn.model_selection import cross_val_score

# 4. Calcular el score con la validación cruzada CV=5

cv_SVM_tunned = cross_val_score(modelo_SVM_tunned, X,y, scoring='neg_mean_squared_error', cv=5)

cv_SVM_tunned
```

```python
# 5. Realizar el cambio de signo y sacar la raiz
rmse_SVM_tunned = np.sqrt(- cv_SVM_tunned)
```

```python
# 6. Mostrar el resultado usando la funcion display_scores

media_SVM_tunned = rmse_SVM_tunned.mean()
std_SVM_tunned = rmse_SVM_tunned.std()

print(media_SVM_tunned, std_SVM_tunned)
```
